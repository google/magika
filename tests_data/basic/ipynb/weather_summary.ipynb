# Fictional weather station summary

This notebook summarizes an original, synthetic set of weather readings for three imaginary stations.

```python
readings = [
    {"station": "Glasswing Ridge", "temperature_c": 2.4, "humidity_pct": 81},
    {"station": "Glasswing Ridge", "temperature_c": 8.7, "humidity_pct": 62},
    {"station": "Sundial Basin", "temperature_c": 9.8, "humidity_pct": 44},
    {"station": "Sundial Basin", "temperature_c": 17.3, "humidity_pct": 29},
    {"station": "Lantern Coast", "temperature_c": 11.5, "humidity_pct": 88},
    {"station": "Lantern Coast", "temperature_c": 13.2, "humidity_pct": 79},
]
```

```python
stations = sorted({row["station"] for row in readings})
mean_temperature = {
    station: round(sum(row["temperature_c"] for row in readings if row["station"] == station) /
                   sum(row["station"] == station for row in readings), 1)
    for station in stations
}
mean_temperature
```
